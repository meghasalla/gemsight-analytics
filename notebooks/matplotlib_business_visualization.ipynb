# Matplotlib Business Visualization Lab

Use this notebook to study each plot independently before running the six-panel dashboard. All data is synthetic.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
data = pd.read_csv(root / 'data' / 'jewelry_monthly_performance.csv', parse_dates=['month'])
data.head()
```

## Line plot: sales over time

```python
monthly = data.groupby('month', as_index=False)['actual_sales_lakh'].sum()
fig, ax = plt.subplots(figsize=(11, 4))
ax.plot(monthly['month'], monthly['actual_sales_lakh'], marker='o')
ax.set(title='Total Jewelry Sales Trend', xlabel='Month', ylabel='Sales (₹ lakh)')
fig.tight_layout()
plt.show()
```

## Scatter plot: footfall versus sales

```python
fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(data['footfall'], data['actual_sales_lakh'], alpha=0.55)
ax.set(title='Footfall vs Sales', xlabel='Monthly visitors', ylabel='Sales (₹ lakh)')
fig.tight_layout()
plt.show()
```
